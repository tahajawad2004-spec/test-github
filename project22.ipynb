# EDA Framework

For future datasets, remember this sequence:

**Understand → Validate → Clean → Describe → Visualize → Compare → Investigate → Explain → Recommend**

A data analyst should not stop at *"sales decreased."* Continue asking:

**Where? → When? → Which segment/product/customer? → What variables are associated with it? → Is it an anomaly or recurring pattern? → What should the business investigate or do next?**

## Step 10 — Detect outliers using IQR

The common IQR rule defines potential outliers as values below:

`Q1 - 1.5 × IQR`

or above:

`Q3 + 1.5 × IQR`


```python
numerical_columns= df.select_dtypes(include="number").columns
outlier_summary = {}
for col_name in numerical_columns:
    if col_name in df.columns:
        # 1. حساب Q1 و Q3 والـ IQR
        Q1 = df[col_name].quantile(0.25)
        Q3 = df[col_name].quantile(0.75)
        IQR = Q3 - Q1
        
        # 2. تحديد الحدود الأدنى والأعلى
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        
        # 3. إيجاد القيم الشاذة وحساب عددها
        outliers = df[(df[col_name] < lower_bound) | (df[col_name] > upper_bound)][col_name]
        outlier_count = outliers.count()
        
        outlier_summary[col_name] = {
            'Q1': Q1,
            'Q3': Q3,
            'IQR': IQR,
            'Lower Bound': lower_bound,
            'Upper Bound': upper_bound,
            'Outliers Count': outlier_count,
            'Percentage (%)': round((outlier_count / len(df)) * 100, 2)
        }
```

## Step 12 — Compare performance across groups

**Ask:** use Groupby

```python
# Convert price to numeric
df["price_numeric"] = pd.to_numeric(
    df["price"]
    .astype(str)
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False),
    errors="coerce"
)

# Group by room type
grouped = df.groupby("room_type").agg({
    "price_numeric": "mean",
    "number_of_reviews": "mean",
    "review_scores_rating": "mean"
})

# Display results
print(grouped)
```

## Step 13 — Time-series EDA

**Ask:** what is trend over time? Are some months stronger than others?


```python
# Make sure last_scraped is a datetime
df["last_scraped"] = pd.to_datetime(df["last_scraped"], errors="coerce")

# Group listings by month
monthly_listings = (
    df.groupby(df["last_scraped"].dt.to_period("M"))
.size()
.reset_index(name="listing_count")
)

# Convert period back to datetime
monthly_listings["last_scraped"] = monthly_listings["last_scraped"].dt.to_timestamp()

print(monthly_listings)

# Visualize the trend
fig = px.line(
    monthly_listings,
    x="last_scraped",
    y="listing_count",
    markers=True,
    title="Number of Listings Over Time",
    labels={
    "last_scraped": "Month",
    "listing_count": "Number of Listings"
    }
)

fig.show()
```

```python
# Monthly average estimated revenue
monthly_revenue = (
    df.groupby(df["last_scraped"].dt.to_period("M"))["estimated_revenue_l365d"]
      .mean()
      .reset_index()
)

monthly_revenue["last_scraped"] = monthly_revenue["last_scraped"].dt.to_timestamp()

fig = px.bar(
    monthly_revenue,
    x="last_scraped",
    y="estimated_revenue_l365d",
    title="Average Estimated Revenue by Month",
    labels={
    "last_scraped": "Month",
    "estimated_revenue_l365d": "Average Estimated Revenue"
    }
)

fig.show()
```

## Step 14 — Find the best and worst performers

A useful EDA habit is to inspect both extremes rather than only averages.


```python
worst_performers = (
    df[["id", "name", "room_type", "price_quote_price_per_night",
    "estimated_occupancy_l365d", "estimated_revenue_l365d"]]
    .sort_values("estimated_revenue_l365d", ascending=True)
    .head(10)
)

print("Bottom 10 Performing Listings:")
worst_performers
```

```python
# Best listings by estimated revenue
best_performers = (
    df[["id", "name", "room_type", "price_quote_price_per_night",
        "estimated_occupancy_l365d", "estimated_revenue_l365d"]]
    .sort_values("estimated_revenue_l365d", ascending=False)
    .head(10)
)
best_performers

print("Top 10 Best Performing Listings:")
best_performers# Worst listings by estimated revenue
```

## Step 15 — Business questions

Try answering these from the analysis above:

1. Which region generates the highest sales?
2. Which region generates the highest profit?
3. Which category generates the highest sales?
4. Which category is most profitable?
5. Does higher discount appear to reduce profitability?
6. Which orders are unusually large?
7. Are high-sales orders always high-profit orders?
8. Which month has the highest sales?
9. Which customer segment contributes the most revenue?
10. What data-quality issues did we discover?


```python
# 1. Which room type has the highest average revenue?
print(
    df.groupby("room_type")["estimated_revenue_l365d"]
      .mean()
      .sort_values(ascending=False)
)

# 2. Which neighbourhood has the highest average revenue?
print(
    df.groupby("neighbourhood_cleansed")["estimated_revenue_l365d"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

# 3. Which property type has the highest average revenue?
print(
    df.groupby("property_type")["estimated_revenue_l365d"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

# 4. Which room type has the highest average occupancy?
print(
    df.groupby("room_type")["estimated_occupancy_l365d"]
      .mean()
      .sort_values(ascending=False)
)

# 5. Does higher price appear to be associated with lower occupancy?
print(
    df[["price_quote_price_per_night", "estimated_occupancy_l365d"]]
      .corr()
)

# 6. Which listings have unusually high estimated revenue?
Q1 = df["estimated_revenue_l365d"].quantile(0.25)
Q3 = df["estimated_revenue_l365d"].quantile(0.75)
IQR = Q3 - Q1

high_revenue = df[
    df["estimated_revenue_l365d"] > Q3 + 1.5 * IQR
]

print(high_revenue[[
    "id", "name", "room_type", "estimated_revenue_l365d"
]].sort_values(
    "estimated_revenue_l365d",
    ascending=False
).head(10))

# 7. Are high-price listings always high-revenue listings?
print(
    df[[
        "price_quote_price_per_night",
        "estimated_revenue_l365d"
    ]].corr()
)

# 8. Which month has the highest average estimated revenue?
print(
    df.groupby(df["last_scraped"].dt.month)["estimated_revenue_l365d"]
      .mean()
      .sort_values(ascending=False)
)

# 9. Which room type contributes the most total estimated revenue?
print(
    df.groupby("room_type")["estimated_revenue_l365d"]
      .sum()
      .sort_values(ascending=False)
)

# 10. Check important data-quality issues
print("Duplicate rows:", df.duplicated().sum())
print("\nMissing values:")
print(df.isnull().sum().sort_values(ascending=False).head(20))
```
